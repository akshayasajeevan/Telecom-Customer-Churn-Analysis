## 3.20 Detecting Outliers

### IQR (Interquartile Range):
<p style="
    font-size: 18px;
    line-height: 1.8;">
IQR is a statistical method used to identify outliers by measuring the range between the first quartile (Q1) and the third quartile (Q3).</p>
<p style="
    font-size: 18px;
    line-height: 1.8;">
It helps us identify values that are unusually low or high compared to the rest of the data.</p>

#### Check the numerical columns

```python
# Select numerical columns
df.select_dtypes(include=['int64', 'float64']).describe()
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Examined the statistical summary of numerical columns to identify unusual values before handling outliers.</p>

#### Calculate IQR boundaries

```python
# Calculate IQR boundaries

for col in ['tenure', 'monthly_charges', 'total_charges']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    print(col)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print()
```

<p style="
    font-size: 18px;
    line-height: 1.8;">The IQR method was used to calculate the lower and upper limits for identifying potential outliers in numerical columns.</p>

#### Count potential outliers

```python
# Count potential outliers

for col in ['tenure', 'monthly_charges', 'total_charges']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(col, ":", len(outliers))
```

```python
# Inspect potential outliers in tenure
df.loc[df['tenure'] > 109.5, ['customer_id', 'tenure']]
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Inspected the potential outliers in the tenure column to check for unusually high values . The IQR method identified 91 unusual values in the tenure column. These values were kept as they may be valid service durations.</p>

#### Inspect Monthly Charges Outliers

```python
# Inspect potential outliers in monthly_charges
df.loc[df['monthly_charges'] > 159.46, ['customer_id', 'monthly_charges']]
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Inspected the potential outliers in the monthly_charges column to check for unusually high values.</p>

## 5.1 Adding New Columns

<p style="
    font-size: 18px;
    line-height: 1.8;">Customers are grouped based on their tenure to understand how long they have been using the service.</p>

#### Create a tenure group

```python
def categorize_tenure(tenure):
    if tenure <= 12:
        return 'New'
    elif tenure <= 24:
        return 'Regular'
    elif tenure <= 48:
        return 'Established'
    else:
        return 'Long-term'

df['tenure_group'] = df['tenure'].apply(categorize_tenure)
```

```python
df[['tenure', 'tenure_group']].head(10)
```

<p style="
    font-size: 18px;
    line-height: 1.8;">A new tenure_group column was created to categorize customers into New, Regular, Established, and Long-term groups based on their service duration.</p>

```python
df['tenure_group'].value_counts()
```

## 5.2 Splitting Columns

#### Create Signup Date Columns

```python
# Extracting Signup Month and Year
df['signup_year'] = df['signup_date'].dt.year
```

```python
df['signup_month'] = df['signup_date'].dt.month_name()
```
