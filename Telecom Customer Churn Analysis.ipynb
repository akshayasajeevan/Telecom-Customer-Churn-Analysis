## 3.20 Detecting Outliers

### IQR (Interquartile Range):
<p style="
    font-size: 18px;
    line-height: 1.8;">
IQR is a statistical method used to identify outliers by measuring the range between the first quartile (Q1) and the third quartile (Q3).</p>
<p style="
    font-size: 18px;
    line-height: 1.8;">
It helps us identify values that are unusually low or high compared to the rest of the data.</p>

#### Check the numerical columns

```python
# Select numerical columns
df[['tenure', 'monthly_charges', 'total_charges']].describe()
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Examined the statistical summary of numerical columns to identify unusual values before handling outliers.</p>

#### Calculate IQR boundaries

```python
# Calculate IQR boundaries

for col in ['tenure', 'monthly_charges', 'total_charges']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    print(col)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print()
```

<p style="
    font-size: 18px;
    line-height: 1.8;">The IQR method was used to calculate the lower and upper limits for identifying potential outliers in numerical columns.</p>

#### Count potential outliers

```python
# Count potential outliers

for col in ['tenure', 'monthly_charges', 'total_charges']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(col, ":", len(outliers))
```

```python
# Inspect potential outliers in tenure
df.loc[df['tenure'] > 109.5, ['customer_id', 'tenure']]
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Inspected the potential outliers in the tenure column to check for unusually high values .</p>

#### Inspect Monthly Charges Outliers

```python
# Inspect potential outliers in monthly_charges
df.loc[df['monthly_charges'] > 159.46, ['customer_id', 'monthly_charges']]
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Inspected the potential outliers in the monthly_charges column to check for unusually high values.</p>

#### Inspect the highest monthly charges

```python
# Display the 10 highest monthly charges
df.nlargest(10, 'monthly_charges')[['customer_id', 'monthly_charges']]
```

<p style="
    font-size: 18px;
    line-height: 1.8;">Examined the highest monthly charges to identify unusually large values.</p>

### Observation:
<p style="
    font-size: 18px;
    line-height: 1.8;">
The IQR method identified 91 potential outliers in the tenure column and 200 potential outliers in the monthly_charges column. These outliers were inspected and removed using the IQR method with the AND (&) operator.</p>

```python
df.nlargest(10, 'monthly_charges')[['customer_id', 'internet_service', 'contract', 'tenure', 'monthly_charges', 'total_charges', 'churn']]
```

```python
# remove outlier

cols = ['tenure', 'monthly_charges']

Q1 = df[cols].quantile(0.25)
Q3 = df[cols].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[
    (df['tenure'] >= lower_limit['tenure']) &
    (df['tenure'] <= upper_limit['tenure']) &
    (df['monthly_charges'] >= lower_limit['monthly_charges']) &
    (df['monthly_charges'] <= upper_limit['monthly_charges'])]
```

```python
for col in ['tenure', 'monthly_charges']:
    outliers = df[(df[col] < lower_limit[col]) | (df[col] > upper_limit[col])]
    print(col, ":", len(outliers))
```

<p style="
    font-size: 18px;
    line-height: 1.8;">The IQR method identified no potential outliers in the total_charges column. Therefore, no further inspection or treatment was required.</p>

### Observation:
<p style="
    font-size: 18px;
    line-height: 1.8;">Potential outliers were identified in the tenure and monthly_charges columns using the IQR method. These outliers were removed using the AND (&) operator. No potential outliers were found in the total_charges column.</p>

## 5.1 Adding New Columns

<p style="
    font-size: 18px;
    line-height: 1.8;">Customers are grouped based on their tenure to understand how long they have been using the service.</p>

#### Create a tenure group

```python
def categorize_tenure(tenure):
    if tenure <= 12:
        return 'New'
    elif tenure <= 24:
        return 'Regular'
    elif tenure <= 48:
        return 'Established'
    else:
        return 'Long-term'

df['tenure_group'] = df['tenure'].apply(categorize_tenure)
```
